```python
# Load the dataset
vehicles_df = pd.read_csv('../vehicles_us.csv')

vehicles_df.head()
```

## 2. Data Cleaning and Correction

```python
# Replace missing values in 'paint_color' with 'Unknown'
vehicles_df['paint_color'] = vehicles_df['paint_color'].fillna('Unknown')

# Replace missing values in 'is_4wd' with 0
vehicles_df['is_4wd'] = vehicles_df['is_4wd'].fillna(0)

# Convert 'is_4wd' to integer type
vehicles_df['is_4wd'] = vehicles_df['is_4wd'].astype(int)

# Change the data type of 'date_posted' to datetime
vehicles_df['date_posted'] = pd.to_datetime(vehicles_df['date_posted'])

vehicles_df.head()
```

```python
# Verify the changes made to the dataset
vehicles_df.info()
```

```python
# Verify if there are duplicates
print(vehicles_df.duplicated().sum(), "duplicates found in the dataset.")
```

```python
# Implicitly duplicates in 'model' column
sorted(vehicles_df['model'].unique())
```

```python
# Remove 'benze' word from 'model' column
vehicles_df['model'] = vehicles_df['model'].str.replace('benze ', '', regex=False)
sorted(vehicles_df['model'].unique())[70:74]
```

```python
# Convert 'type' in lower case
vehicles_df['type'] = vehicles_df['type'].str.lower()
sorted(vehicles_df['type'].unique())
```

```python
# Implicitly duplicates in 'paint_color' column
sorted(vehicles_df['paint_color'].unique())
```

## 3. Data Enrichment

```python
# Split Manufacturer and Model into separate columns
vehicles_df[['manufacturer', 'model']] = vehicles_df['model'].str.split(' ', n=1, expand=True)
vehicles_df.head()
```

**Findings**

* **Missing values** in the model_year, cylinders, and odometer columns were **retained.** During subsequent stages of data processing, these records will be filtered as needed to prevent them from affecting calculations while preserving potentially valuable information.

* The **data types** of the columns were reviewed and **corrected** to ensure consistency and support accurate analysis.

* The entry "mercedes-benz benze sprinter 2500" appears to contain a typographical error. While Mercedes-Benz is the vehicle manufacturer and Sprinter 2500 is a recognized model, the term **"benze"** does not correspond to any known Mercedes-Benz model and is likely the result of inconsistent or **erroneous data** entry.

* The model column was split into separate fields because it contained both the **manufacturer** name and the vehicle **model**, allowing for more detailed analysis by brand and model.

* **No** explicit or implicit **duplicate** records were identified in the dataset.

* **Missing values** in the paint_color column were **replaced** with **"Unknown"**, while missing values in the is_4wd column were replaced with **0**, indicating that four-wheel drive was not specified for those records.

## 4. Exploratory Data Analysis

```python
# Vehicles types by manufacturer
manufacturer_type_counts = vehicles_df.groupby(['manufacturer', 'type']).size().reset_index(name='count')

# Order manufacturers by total count
manufacturer_list = manufacturer_type_counts.groupby('manufacturer')['count'].sum().sort_values(ascending=False).index.to_list()

# Create a stacked bar chart using Plotly Express
fig = px.bar(
    manufacturer_type_counts,
    x='manufacturer',
    y='count',
    color='type',
    barmode='stack',
    category_orders={'manufacturer': manufacturer_list},
)

# Title and axis labels
fig.update_layout(
    title='Vehicle Types by Manufacturer',
    xaxis_title='Manufacturer',
    yaxis_title='Count',
    legend_title='Vehicle Type',
    #xaxis_tickangle=-45,
)

fig.show()
```

```python
# Kilometers vs Price scatter plot
fig = px.scatter(
    vehicles_df[vehicles_df['type'] == 'sedan'],  # Filter outliers for better visualization
    x='odometer',
    y='price',
    title='Kilometers vs Price'
)

# Title and axis labels
fig.update_layout(
    xaxis_title='Odometer (km)',
    yaxis_title='Price ($)',
)

fig.show()
```

```python
# Correlation between 'odometer' and 'price' sedan type vehicles
vehicles_df[vehicles_df['type'] == 'sedan'][['odometer', 'price']].corr()
```

```python
# Histogram of model year by condition
fig = px.histogram(
    vehicles_df,
    x='model_year',
    title='Distribution of Vehicle Years',
    color='condition',
)

# Title and axis labels
fig.update_layout(
    xaxis_title='Model Year',
    yaxis_title='Count',
)

fig.show()
```
